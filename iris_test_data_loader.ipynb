# Import Required Libraries
This section imports pandas and scikit-learn for loading and manipulating the iris dataset.

```python
# Import pandas and scikit-learn
import pandas as pd
from sklearn import datasets
```

# Load the Iris Test Data
This section loads the iris dataset using scikit-learn and converts it to a pandas DataFrame.

```python
# Load the iris dataset from sklearn
dataset = datasets.load_iris()
iris_df = pd.DataFrame(data=dataset.data, columns=dataset.feature_names)
iris_df['target'] = dataset.target
```
