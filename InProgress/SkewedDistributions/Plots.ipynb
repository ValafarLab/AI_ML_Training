```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import gamma

# Parameters for the gamma distribution
mean = 80
std_dev = 10
size = 10000
shape_param = 2  # Shape parameter for gamma distribution (affects skewness)

# Generate a left-skewed distribution by reflecting a right-skewed distribution
right_skewed_data = gamma.rvs(a=shape_param, scale=std_dev / np.sqrt(shape_param), size=size)
left_skewed_data = -right_skewed_data

# Rescale to match desired mean and standard deviation
left_skewed_data = (left_skewed_data - np.mean(left_skewed_data)) / np.std(left_skewed_data) * std_dev + mean

# Define the PDF for the reflected gamma distribution
x_vals = np.linspace(min(left_skewed_data), max(left_skewed_data), 1000)
right_pdf = gamma.pdf(-x_vals, a=shape_param, scale=std_dev / np.sqrt(shape_param))
left_pdf = right_pdf  # Reflect the PDF

# Plot the continuous distribution
plt.figure(figsize=(8, 5))
plt.plot(x_vals, left_pdf, color='lightcoral', lw=2)
plt.title('Continuous Left-Skewed Distribution (Mean = 80, Std Dev = 10)')
plt.xlabel('Value')
plt.ylabel('Density')
plt.grid(alpha=0.3)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters for the normal distribution
mean = 50
std_dev = 10
size = 10000  # Number of data points

# Generate the normal distribution
data = np.random.normal(loc=mean, scale=std_dev, size=size)

# Plot the distribution
plt.figure(figsize=(8, 5))
plt.hist(data, bins=50, color='skyblue', edgecolor='black', alpha=0.7, density=True)
plt.title('Normal Distribution (Mean = 50, Std Dev = 10)')
plt.xlabel('Value')
plt.ylabel('Density')
plt.grid(alpha=0.3)

# Add a line for the theoretical distribution
x = np.linspace(mean - 4*std_dev, mean + 4*std_dev, 1000)
y = (1 / (std_dev * np.sqrt(2 * np.pi))) * np.exp(-0.5 * ((x - mean) / std_dev) ** 2)
plt.plot(x, y, color='red', lw=2, label='Theoretical PDF')

plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters for the normal distribution
mean = 80
std_dev = 10
size = 10000  # Number of data points

# Generate the normal distribution
data = np.random.normal(loc=mean, scale=std_dev, size=size)

# Apply a transformation to introduce left skewness
left_skewed_data = np.power(data - np.min(data) + 1, 0.3)  # Ensure non-negativity before applying the power
left_skewed_data = (left_skewed_data - np.mean(left_skewed_data)) / np.std(left_skewed_data) * std_dev + mean

# Plot the left-skewed distribution
plt.figure(figsize=(8, 5))
plt.hist(left_skewed_data, bins=50, color='lightcoral', edgecolor='black', alpha=0.7, density=True)
plt.title('Left-Skewed Distribution (Mean = 50, Std Dev = 10)')
plt.xlabel('Value')
plt.ylabel('Density')
plt.grid(alpha=0.3)

# Display the plot
plt.show()
```

```python

import numpy as np
import matplotlib.pyplot as plt

# Define a simple parton distribution function (PDF) model
def pdf(x, A=1, alpha=0.1, beta=1):
    """
    A simple power law for the parton distribution function
    where x is the momentum fraction of the parton
    A, alpha, beta are parameters controlling the shape of the distribution.
    """
    return A * (x**alpha) * (1 - x)**beta

# Generate x values (momentum fraction)
x = np.linspace(0.01, 1, 100)  # Avoiding 0 and 1 to prevent singularities

# Calculate PDF for each x
y = 6*pdf(x, A=1, alpha=8, beta=2)

# Plot the PDF
plt.figure(figsize=(8, 6))
plt.plot(x, y, label=r'$x^{\alpha} (1-x)^{\beta}$', color='blue')
plt.title('Parton Distribution Function (PDF)')
plt.xlabel(r'$\ x$ (momentum fraction)')
plt.ylabel('PDF')
plt.grid(True)
plt.legend()
plt.show()
```
