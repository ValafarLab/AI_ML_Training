```python
from transformers3d import GPT2Tokenizer
```

```python
# Load a tokenizer (e.g., GPT-2 tokenizer)
tokenizer = GPT2Tokenizer.from_pretrained("gpt2")
```

```python
# The input statement
input_text = "provide a summary of NMR to a high school student"
```

```python
# Tokenize the input text
tokens = tokenizer.tokenize(input_text)
token_ids = tokenizer.convert_tokens_to_ids(tokens)
```

```python
# Output the tokens and their corresponding IDs
print("Tokens:", tokens)
print("Token IDs:", token_ids)
```
