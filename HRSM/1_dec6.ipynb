```python
from sklearn.cluster import KMeans
```

```python
lat_lon_df = df[['latitude','longitude','price']]
kmeans_a = KMeans(n_clusters=5)
cluster_ids = kmeans_a.fit_predict(lat_lon_df)
lat_lon_df['cluster'] = cluster_ids
sns.scatterplot(data=lat_lon_df,x='latitude',y='longitude',hue="cluster",s=5)
```
