```python
import pandas as pd
import os
```

```python
# Folder where your Excel files are stored
# folder_path = "/home/homayoun/Dropbox/Projects/VisualCode/Python/CAE/New/Data/"
folder_path = os.path.realpath('./Data/')
#folder_path = os.path.dirname(os.path.realpath('./New/Data/'))
print(folder_path)
# Range of years you want to include
years = range(2014, 2025)  # includes 2024

# List to hold data from each file
data_frames = []

for year in years:
    file_name = f"{year}-Terminal.xlsx"
    file_path = os.path.join(folder_path, file_name)
    print(file_path)

    if os.path.exists(file_path):
        print(f"Reading {file_name}...")
        df = pd.read_excel(file_path)
#        print(df.head())
        df["Year"] = year  # optional: add year column
        data_frames.append(df)
    else:
        print(f"⚠️ File not found: {file_name}")

# Combine all dataframes into one
if data_frames:
    combined_df = pd.concat(data_frames, ignore_index=True)

    # Save to Excel or CSV
    combined_df.columns= ['MeterID','DateTime','KVARH53','KWH32',
                 'KVARH42','KVAH52','PF62','KWH11','KVARH21',
                 'KVAH71','PF81','KW','Year']
    #combined_df.to_excel(os.path.join(folder_path, "Combined_Terminal.xlsx"), index=False)
    #combined_df.to_excel(os.path.join(folder_path, "Combined_Terminal.xlsx"), index=False)
    combined_df[['DateTime', 'KWH11', 'KVAH71', 'KW', 'Year']].to_csv(os.path.join(folder_path, "Combined_Terminal.csv"), index=False)
    combined_df[['DateTime', 'KWH11', 'KVAH71', 'KW', 'Year']].to_csv(os.path.join(folder_path, "Combined_Terminal.csv"), index=False)

    print("✅ All files combined successfully!")
else:
    print("No data combined. Please check your file names and folder path.")
```
